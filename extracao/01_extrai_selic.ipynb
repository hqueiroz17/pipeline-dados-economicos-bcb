```python
import boto3

s3 = boto3.client(
    "s3",
    aws_access_key_id=access_key,
    aws_secret_access_key=secret_key,
    region_name="sa-east-1"
)

buckets = s3.list_buckets()
for b in buckets["Buckets"]:
    print(b["Name"])

```

```python
import requests
import json

series = {
    "selic": "11",
    "ipca": "433",
    "dolar": "1"
}

data_inicial = "01/01/2024"
data_final = "13/09/2026"

dados_completos = {}

for nome, codigo in series.items():
    url = f"https://api.bcb.gov.br/dados/serie/bcdata.sgs.{codigo}/dados?formato=json&dataInicial={data_inicial}&dataFinal={data_final}"
    resposta = requests.get(url)
    dados_completos[nome] = resposta.json()
    print(f"{nome}: {len(dados_completos[nome])} registros")
```

```python
from datetime import datetime

data_hoje = datetime.today().strftime("%Y-%m-%d")

for nome, conteudo in dados_completos.items():
    nome_arquivo = f"bronze/{data_hoje}/{nome}.json"
    s3.put_object(
        Bucket="pipeline-economico-hugoqueiroz",
        Key=nome_arquivo,
        Body=json.dumps(conteudo)
    )
    print(f"Salvo: {nome_arquivo}")
```
